```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
```

```python
x = np.array([[0],[2],[3],[7],[12]], float)
fig, ax = plt.subplots(1, 2, figsize=(10,4))
for a, m in zip(ax, ['single','complete']):
    L = linkage(x, method=m)
    print(m, L[:,2])
    dendrogram(L, labels=[0,2,3,7,12], ax=a)
    a.set_title(m)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
```

```python
x = np.array([1, 2, 3, 8, 9, 10, 25], dtype=float)

mu = x.mean()
sigma = x.std(ddof=0)

z_25 = (25 - mu) / sigma

print("Mean:", mu)
print("Population standard deviation:", sigma)
print("z-score of 25:", z_25)
print("Flagged by |z| > 3:", abs(z_25) > 3)
```
